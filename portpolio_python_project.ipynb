**Task 4: Data cleaning**

```python
# Dropping all the missing values

data.dropna(inplace = True)

# data.fillna()
data.isnull().sum()
```

```python
data.shape
```

```python
# dealing with duplicate rows
data.duplicated().sum()

# deleating all duplicate data
#data[data.duplicated()]

data.drop_duplicates(inplace=True)
data.duplicated().sum()


```

```python
# type casting
# changing data types
data.dtypes

data['id'] = data['id'].astype(object) 
data.dtypes

data['host_id'] = data['host_id'].astype(object)
data.dtypes
# i did this changes for output below coding because the output did not changed as showing in the youtube so so idi all
 
data['name'] = data['name'].astype(object)
data.dtypes

data['host_name'] = data['host_name'].astype(object)
data.dtypes

data['neighbourhood_group'] = data['neighbourhood_group'].astype(object)
data.dtypes

data['neighbourhood'] = data['neighbourhood'].astype(object)
data.dtypes


data['room_type'] = data['room_type'].astype(object)
data.dtypes

data['last_review'] = data['last_review'].astype(object)
data.dtypes



data['license'] = data['license'].astype(object)
data.dtypes

data['rating'] = data['rating'].astype(object)
data.dtypes

data['bedrooms'] = data['bedrooms'].astype(object)
data.dtypes

data['baths'] = data['baths'].astype(object)
data.dtypes

```

**Task 5: Data Analysis**

**Univariate Analysis**

```python
# identifying  outliers in price
df= data[data['price'] <1500]
sns.boxplot(data=df,x='price')
```

```python
# price distribution

plt.figure(figsize=(8,5))
sns.histplot(data=df, x='price',bins=100)
plt.title('Price Distribuition')
plt.ylabel("frequency")
```

```python
df.dtypes
```

```python
# price distribution

plt.figure(figsize=(6,3))
sns.histplot(data=df, x='availability_365')
plt.title('availability_365 Distribuition')
plt.ylabel("frequency")
plt.show()
```

```python
# Average price per bed
df.groupby(by='neighbourhood_group')['price per bed'].mean()
```

```python
# Price dependency on neighbourhood
sns.barplot(data=df,x='neighbourhood_group',y='price',hue='room_type')
```

```python
sns.pairplot(data=df,vars=['price','minimum_nights','number_of_reviews','availability_365'],hue='room_type')
```

```python
#Heat map- correlation  of one variable with others for numerical column

corr=df[['latitude','longitude','price','minimum_nights','number_of_reviews','reviews_per_month','availability_365','beds']].corr()
corr
plt.Figure(figsize=(8,6))
sns.heatmap(data=corr,annot=True)
```
